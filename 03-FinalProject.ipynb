## Abstract

#### Background and Our Motivation
This project investigates whether or not the increased use of streaming services (after 2013) has led to differences in the variability of audio features in mainstream popular music. We were interested in understanding how the degree of homogenization (i.e. the relative similarity/variability of songs in a given year) in the pre-streaming era (2000-2012) versus the post-streaming era (2013-2026) changed, if at all. We aimed to conduct an EDA particularly for mainstream popular music, quantified by songs in these eras that were on the Billboard Hot 100 chart. Prior discussions about streaming’s effect on music suggest that it may have led to popular songs getting shorter and possibly more similar, which motivated us to explore whether popular music really has changed in a meaningful way. With streaming platforms and algorithmic recommendations playing a larger role in music distribution and discovery in the post-streaming era (2013-2026), we aim to determine whether or not the prevalence of streaming, and the measure of a song's 'success' by its number of streams, has influenced the variability (or lack thereof) of song features found in popular music.
#### Data and Methods
To explore this question, we combined multiple datasets including Billboard Hot 100 chart rankings and Spotify audio feature metadata. From the Billboard data, we extracted the top fifty songs from each year and matched them with Spotify’s audio features such as danceability, energy, valence, acousticness, instrumentalness, tempo, loudness, and speechiness. We performed an EDA to examine the variability for each feature between pre-streaming and post-streaming eras. We also computed an overall musical diversity score for both eras, which aggregated the feature variabilities into a single comparable value so we could determine whether mainstream music in one era had an increases or decrease in variability compared to the other.
#### Major Findings
Our findings from our conducted research suggest that while several musical features have gone through a statistically significant change in the post-streaming era, such as increased energy, louder production, shorter song durations, and increased explicitness, these changes are not indicative of a major shift in the overall variability of musical features among popular music. The variability across many features remains relatively broad, with very similar musical diversity scores (0.7098 vs. 0.7085), suggesting that while streaming platforms and algorithms may influence popular music trends, they don’t necessarily lead to a complete homogenization of mainstream music.
#### Conclusion and Implications
Overall, our analysis provides insight into how new technology and the changes that come from adopting new technology can shape the sound of popular music over time. Since the diversity of popular music has mostly stayed the same, with only some features having a marked change such as song duration and explicitness, we can conclude that popular music continues to maintain a reasonable amount of musical diversity, even throughout the overhaul that was algorithm-driven streaming. Though this relationship between popular music and musical diversity can definitely be further explored in different ways (such as by doing a deeper NLP analysis of song lyrics, or doing focused analyses of individual genres), a preliminary explanation for the unchanging nature of musical diversity could be indicative of the fact that society has maintained diverse tastes, and it makes sense that popular music has also remained diverse, as it is society that drives popular music.

## Background and Prior Work

As digitalization has rapidly advanced, no stone has been left unturned. The way we listen to music has been one of the biggest changes, progressing from physical cassettes, records, and CDs to singular devices that hold hundreds of millions of tracks. This unparalleled access to music has not come lightly. Ever since digitalization reached music in the 2010s, consumers have been able to purchase or subscribe to a streaming platform to gain access to a wide variety of songs readily available to them. Since the release of streaming platforms like Spotify and Apple Music, the barrier of entry to consume and release music has lowered exponentially.

Perhaps one of the clearest signs of how music has changed is the change in the average length of music over time, which has decreased by more than 30 seconds. Many of the top songs of the past few years have clocked in at under three minutes, a clear departure from longer music. This is most likely attributable to the differences streaming makes in song structure. According to music information scientists at Spotify, Nicola Montecchio, Pierre Roy, and François Pachet, there is a correlation between skipping and song structure: users are more likely to skip songs before they have reached 30 seconds. Since streaming services only pay artists for “plays” and “plays” only count if a song has been played for longer than 30 seconds, it stands to reason that many artists and songwriters have focused their energy towards creating catchy music that’s interesting enough to capture attention in the beginning, but short enough that listeners won’t get bored and skip too early, leaving artists without their hard-earned revenue.

A study by NYU PhD Alumni in the Wall Street Journal supports the Spotify scientist's findings, sharing how streaming has encouraged the practice of shortening songs to  catch the listener's attention as quickly as possible. They cite an interview with Grammy-winning producer and performer Mark Ronson to highlight the reason behind the practice: “All your songs have to be under three minutes and 15 seconds because if people don’t listen to them all the way to the end, they go into this ratio of ‘non-complete heard,’ which sends your Spotify rating down,” (Ronson in Rosenblatt). They further build on this phenomenon by highlighting how, though song length has decreased, album length has increased. This trend has been seen largely because when an artist releases a new album, fans often listen to the entire collection, generating more revenue per song. The widespread nature of this practice leads us to theorize that song structure has become a notably more homogenized aspect of music due to streaming's influence. 

In a previous study from Tilburg University, the impact of streaming services on homogenized mainstream music is mixed. Researchers Knox and Datta look at the supply and demand side of the industry, focusing on the “acoustic diversity” of a given song: the pitch, loudness, valence, tempo, and duration. In the early beginnings of platforms like Napster, an illegal music sharing platform, YouTube, and Spotify, artists were able to easily create and distribute their music to a large audience in a way that had not been seen before. Musicians tended to mimic each other in the early stages of the new widespread distribution of music, leading to an initial decrease in diversity in music quality. However, when streaming services like Spotify became more popular, algorithms in place were able to provide new recommendation systems and search tools that enabled listeners to discover a vast amount of music readily available to them. From a demand standpoint, consumers have more choices of music selection available to them, thus allowing for more diversity within their consumption. Thus, despite its initial dip, Knox and Datta's study found a weak, steady increase in acoustic diversity since the release of streaming platforms due to the large catalog of recorded music available. 

Overall, this research highlighted how recommendation algorithms and user skips have become imperative considerations in the music-making process. The more a song appeals to an algorithm or the widest audience possible, the more streams a song has, and the more money it makes. In this project, we’d like to explore whether or not these factors have led to music becoming more homogenized from the pre-streaming (2000-2012) to the post-streaming (2013-2026) eras. Our definition of homogeneity will include several features, including acousticness, danceability, energy, instrumentalness, loudness, speechiness, tempo, valence, key, explicitness, duration, and lyric sentiment. Energy measures the "danceability" and animation of a given song; faster tempos and louder volumes connote higher scales of danceability and activity in a track, whereas softer, quieter volumes and slower tempos connote lower scales of energy. Valence describes the musical positivity; higher scales of valence include upbeat, happier, and euphoric music, while lower values will include more moody, sadder, and tense musical qualities. High acousticness will include songs that sound organically produced physically by humans with instruments, and lower values of acousticness will include songs that sound electronically produced, including techno-music. Instrumentalness highlights how many words are in a given track. A song with many lyrics will be graded lower on the instrumentalness scale, whereas tracks with mostly instrumental portions with very few lyrics and vocals will be graded higher on the scale. Lyric sentiment will be classified into positive and negative categories by the NLTK VADER tool. We aim to measure the standard deviation and variance of these values within both time epochs in order to determine homogeneity differences.

```python
import requests
import pandas as pd
import time
from io import StringIO

start_year = 2000
end_year = 2025

all_years = []

headers = {
    "User-Agent": "Mozilla/5.0"
}

for year in range(start_year, end_year + 1):
    try:
        url = f"https://en.wikipedia.org/wiki/Billboard_Year-End_Hot_100_singles_of_{year}"
        
        response = requests.get(url, headers=headers)
        response.raise_for_status()

        html = StringIO(response.text)
        tables = pd.read_html(html)

        for table in tables:
            if "Title" in table.columns or "Song" in table.columns:
                df = table
                break

        df = df.head(10).copy()

        df = df.iloc[:, -2:]
        
        df.columns = ["song", "artist"]
        
        df["year"] = year
        
        df = df[["year", "song", "artist"]]
        
        df.columns = [str(col).lower() for col in df.columns]

        if "title" in df.columns:
            df = df.rename(columns={"title": "song"})
        if "artist(s)" in df.columns:
            df = df.rename(columns={"artist(s)": "artist"})

        if "song" not in df.columns:
            df.rename(columns={df.columns[1]: "song"}, inplace=True)
        if "artist" not in df.columns:
            df.rename(columns={df.columns[2]: "artist"}, inplace=True)

        all_years.append(df)

        print(f"Finished {year}")

    except Exception as e:
        print(f"Skipped {year}: {e}")

final_df = pd.concat(all_years, ignore_index=True)

print(final_df.head())
```

```python
#artist cleanup (will be applied to Spotify 160k later as well)
import re

def normalize_artists(artist_string):
    if pd.isna(artist_string):
        return artist_string
    artist_string = artist_string.lower()
    artist_string = re.sub(r'featuring|feat\.?|with|&| x | and ', ',', artist_string) 
    artist_string = re.sub(r'\s*,\s*', ',', artist_string)
    artist_string = re.sub(r',+', ',', artist_string)
    artist_list = artist_string.split(',')
    artist_list = [a.strip() for a in artist_list if a.strip()]
        
    return ",".join(artist_list)
```

The 160K Spotify Songs (1921–2020) is a large music metadata collection derived from Spotify’s Web API. It contains roughly 160,000 tracks and includes artist information, genre labels, release year, and several audio features.

The most important metrics are Spotify’s audio features, which are standardized numeric descriptors. Danceability (unitless, 0–1) estimates how suitable a track is for dancing. Values closer to 1 indicate more danceable songs. Energy (0–1) measures intensity and activity; higher values correspond to louder and faster tracks. Valence (0–1) describes musical positivity. Higher values sound happier or more euphoric, while lower values tend to sound sadder or more tense. Acousticness (0–1) estimates the likelihood a track is acoustic rather than electronically produced. Instrumentalness (0–1) predicts whether a track contains vocals; values above ~0.5 suggest mostly instrumental music. These features are probabilities or model outputs rather than physical measurements, so they are relative indicators rather than exact physical quantities.

Several features are expressed in real physical units. Tempo is measured in beats per minute (BPM); most contemporary pop songs fall roughly between 90–130 BPM, while extreme values (e.g., >200 BPM) indicate very fast genres like speed metal or certain electronic styles. Duration is given in milliseconds; dividing by 1000 converts to seconds. Loudness is measured in decibels (dB), typically ranging from about −60 dB (very quiet) to 0 dB (maximum digital level). Key is encoded as an integer 0–11 representing the 12 pitch classes (C through B), and mode indicates major (1) or minor (0). Speechiness (0–1) measures the presence of spoken words; values above ~0.66 are likely spoken word (e.g., podcasts, rap verses heavy in speech), while values below ~0.33 are more purely musical. These metrics allow analysis of musical style, mood, production trends, and historical changes over time.

The primary concern with bias in this dataset is derived from the fact that the data is taken from Spotify’s catalog and recommendation algorithms, meaning it reflects Spotify’s licensing agreements and popularity trends, not a perfectly representative sample of all music ever recorded. Earlier decades (e.g., 1920s–1950s) are likely underrepresented and skewed toward recordings that have been digitized and re-released, which may also show biases.
```python
import pandas as pd
import re
import ast

def clean_title(title):
    if pd.isna(title):
        return title
    
    title = title.lower()
    
    # remove quotes
    title = title.replace('"', '')
    title = title.replace("'", "")
    
    # remove parentheses content
    title = re.sub(r"\(.*?\)", "", title)
    
    # remove dash versions
    title = re.sub(r"-.*", "", title)
    
    # remove punctuation
    title = re.sub(r"[^\w\s]", "", title)
    
    return title.strip()

def clean_spotify_artists(artist_string):
    if pd.isna(artist_string):
        return None
    
    # convert string list → python list
    artist_list = ast.literal_eval(artist_string)
    
    # lowercase and strip
    artist_list = [a.lower().strip() for a in artist_list]
    
    return ",".join(artist_list)
```

### Dataset #3: Genius Lyrics Dataset

For this project, we sourced lyrics from the Genius Lyrics Database, filtering entries to match our selected sample of Billboard songs. The Genius database is a crowdsourced repository containing over 8 million lyrics, making it the largest publicly available resource for lyrical and musical knowledge. Contributions are often verified by the artists themselves, ensuring a high degree of accuracy not only for the lyrics but also for the annotations and analyses associated with each song. Using the LyricsGenius Python client, we retrieved lyrics corresponding to our sample of songs from both the pre- and post-streaming eras, enabling a consistent and structured dataset for analysis. These lyrics form the basis for our exploration of lyrical features across the eras of interest, focusing primarily on emotional sentiment. 

https://genius.com/api-clients

```python
import os
import pandas as pd
import re
os.environ["GENIUS_API_TOKEN"] = ""

print("Genius API token set successfully!")
```

```python
#run pip install lyricsgenius first if you do not have it
import lyricsgenius

GENIUS_TOKEN = os.getenv("GENIUS_API_TOKEN")

if GENIUS_TOKEN is None:
    raise ValueError(
        "Genius API token not found. Please set GENIUS_API_TOKEN as an environment variable."
    )

genius = lyricsgenius.Genius(
    GENIUS_TOKEN,
    skip_non_songs=True,
    remove_section_headers=True,
    excluded_terms=["(Remix)", "(Live)"]
)

print("Genius client initialized successfully.")
```

```python
def clean_artist(artist):
    return artist.split(",")[0].strip()
def clean_title(title):
    title = re.sub(r"\(.*?\)", "", title)  
    title = re.sub(r"- .*", "", title)     
    return title.strip()
    
pre_streaming_data = pd.read_csv("data/02-processed/pre_streaming_top10_annual")
post_streaming_data = pd.read_csv("data/02-processed/post_streaming_top10_annual")

pre_streaming_data["genius_artist"] = pre_streaming_data["artists"].apply(clean_artist)
pre_streaming_data["genius_title"] = pre_streaming_data["song"].apply(clean_title)

post_streaming_data["genius_artist"] = post_streaming_data["artists"].apply(clean_artist)
post_streaming_data["genius_title"] = post_streaming_data["song"].apply(clean_title)
```

```python
pre_streaming_data
```

```python
from concurrent.futures import ThreadPoolExecutor, as_completed
import pandas as pd

def fetch_lyrics(row, era):
    title = row["genius_title"]
    artist = row["genius_artist"]

    try:
        song = genius.search_song(title, artist, get_full_info=False)

        return {
            "title": row["genius_title"],
            "artist": row["artists"],
            "lyrics": song.lyrics if song else None,
            "era": era,
            "year": row["year"]  # 👈 ADD THIS
        }

    except Exception as e:
        print(f"Error with {title}: {e}")
        return {
            "title": row["genius_title"],
            "artist": row["artists"],
            "lyrics": None,
            "era": era,
            "year": row["year"]  # 👈 ADD HERE TOO
        }


lyrics_rows = []

# combine datasets so we only parallelize once
tasks = []

for _, row in pre_streaming_data.iterrows():
    tasks.append((row, "pre"))

for _, row in post_streaming_data.iterrows():
    tasks.append((row, "post"))

# run parallel requests
with ThreadPoolExecutor(max_workers=10) as executor:
    futures = [executor.submit(fetch_lyrics, row, era) for row, era in tasks]

    for future in as_completed(futures):
        lyrics_rows.append(future.result())

# create dataframe
final_df = pd.DataFrame(lyrics_rows)
```

The following section runs the Genius Lyrics through the AI Sentiment Analysis tool created by Nyckel. 

```python
genius_df = pd.read_csv("data/02-processed/genius-lyrics-dataset")
genius_df.dropna()
genius_df = genius_df.loc[:, ~genius_df.columns.str.contains('^Unnamed')]
```

```python
import pandas as pd
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer

nltk.download('vader_lexicon')

sia = SentimentIntensityAnalyzer()

def clean_lyrics(text):
    if pd.isna(text):
        return ""
    return text.replace("\n", " ").strip()

genius_df['lyrics_clean'] = genius_df['lyrics'].apply(clean_lyrics)

genius_df['sent_scores'] = genius_df['lyrics_clean'].apply(sia.polarity_scores)

genius_df['compound'] = genius_df['sent_scores'].apply(lambda x: x['compound'])

def label_sentiment(score):
    if score >= 0.05:
        return "Positive"
    elif score <= -0.05:
        return "Negative"
    else:
        return "Neutral"

genius_df['sentiment'] = genius_df['compound'].apply(label_sentiment)
```

At this stage, we predict that with the exception of duration, energy, and loudness, several of the other features are unlikely to prove a statistically significant difference. We continue with Cohen's analysis of effect size to support or disprove this hypothesis. 

```python
#Cohen's d to measure effect size
def cohens_d(x, y):
    
    nx = len(x)
    ny = len(y)
    
    pooled_std = np.sqrt(((nx-1)*np.var(x) + (ny-1)*np.var(y)) / (nx+ny-2))
    
    return (np.mean(x) - np.mean(y)) / pooled_std
```

```python
#Cohen's d and T-test for continuous features between pre- and post-streaming eras
results = []

for feature in continuous_features:
    
    pre = pre_streaming_features[feature].dropna()
    post = post_streaming_features[feature].dropna()
    
    t, p = stats.ttest_ind(pre, post, equal_var=False)
    
    d = cohens_d(pre, post)
    
    results.append({
        "feature": feature,
        "t_stat": t,
        "p_value": p,
        "cohens_d": d
    })

import pandas as pd
results_df = pd.DataFrame(results)

results_df
```

Similarly, loudness increased significantly (t = 2.35, p = 0.0199, Cohen’s d = 0.323), indicating that modern popular songs are produced with louder levels. These results are consistent with the boxplots, which show an upward shift in the distributions of both energy and loudness after 2013, or rather, after the pre-streaming era. The changes in loudness and energy can be attributed to the evolution in music production technology, particularly better lossless compression and recording technology, which are able to preserve more of the essence of the original recorded music. 

Song duration also changed substantially between eras. Duration decreased significantly in the streaming era with t = 5.06, p < 0.000001, and Cohen’s d = 0.696, representing the largest observed effect size among the tested variables. In contrast, several other features don't show much evidence of change. The above visualizations suggest a potentially signifigant difference in explicitness and in certain keys. In order to confirm, we perform a Chi squared test.

```python
#Chi squared test to measure statistical significance of differences between categorical features
cat_results = []

for feature in categorical_features:
    
    table = pd.crosstab(merged_df[feature], merged_df["era"])
    
    chi2, p, dof, expected = chi2_contingency(table)
    
    cat_results.append({
        "feature": feature,
        "chi2": chi2,
        "p_value": p
    })

cat_results_df = pd.DataFrame(cat_results)
cat_results_df
```

```python
def cramers_v(table):
    
    chi2 = chi2_contingency(table)[0]
    n = table.sum().sum()
    
    r, k = table.shape
    
    return np.sqrt(chi2 / (n * (min(r-1, k-1))))
```

```python
#Using Cramér's V to measure the effect size of the difference between categorical features
for feature in categorical_features:
    
    table = pd.crosstab(merged_df[feature], merged_df["era"])
    
    v = cramers_v(table)
    
    print(feature, v)
```

The proportion of songs marked as explicit increased significantly in the streaming era, according to the Chi-squared test(χ² = 9.00, p = 0.0027), suggesting a greater prevalence of explicit lyrical content in modern popular music, which can be explained by a shift in cultural norms and the societal acceptance of explicit language and explicit themes. However, the distributions of musical key (χ² = 13.03, p = 0.291) and mode (χ² = 0.013, p = 0.910) show no statistically significant differences across eras, indicating that tonal structure has remained stable, likely representing unchanging industry standards and the stability of mainstream tastes.

However, the relatively low Cramér's V values indicate that the strength of association between era and our categorical data values is very little, as generally values above 0.25 are considered significant correlation. Even for explicitness, the Cramér's V value was less than this specified threshold, indicating a relatively weak relationship between explicitness and era, even though it measured to be statistically signifigant by the Chi-squared test.

```python
#Testing normality of results
stats.shapiro(pre_div["diversity_score"])
stats.shapiro(post_div["diversity_score"])
```

```python
#p > 0.05 - results are normal. Using the Welch t-test for normal data:
stats.ttest_ind(
    pre_div["diversity_score"],
    post_div["diversity_score"],
    equal_var=False
)
```

```python
#Effect size of diversity scores
def cohens_d(x,y):
    pooled_sd = np.sqrt((x.var() + y.var()) / 2)
    return (x.mean() - y.mean()) / pooled_sd

cohens_d(pre_div["diversity_score"], post_div["diversity_score"])
```

We conduct several more tests to confirm our hypothesis about a lack of significant difference between our two eras of interest. We therefore conducted a two-sample t-test, which yielded a p-value of 0.946, providing no evidence of a significant difference in mean feature variability between the two eras.

To further quantify the magnitude of this difference, we compute Cohen’s d, obtaining a value of around 0.07, which corresponds to a negligible effect size. This suggests that any observed differences in homogeneity are not only statistically insignificant but also practically minimal.

Taken together, these results indicate that the overall level of musical homogeneity across a wide range of audio characteristics has remained stable despite the transition to the streaming era.

The above Ttest result shows a p-value of over 0.6, which is far above the <=0.05 needed to determine statistical signifigcance. Thus we fail to come to the conlcusion that there is a notable difference between sentiment between songs from the pre and post streaming eras.

However, this data only shows the similarities and differences between the sentiments of both eras. Though the p-value indicated that there is not a statistically significant differnce between the means of sentiment between both eras, there still might be a difference in overall homogeneity. We measure homogeneity based on the variance of compound sentiment values amongst the pre and post streaming eras individually, then compare.  

```python
pre_var = genius_df[genius_df['era'] == 'pre']['compound'].var()
post_var = genius_df[genius_df['era'] == 'post']['compound'].var()

print("Pre variance:", pre_var)
print("Post variance:", post_var)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(data=genius_df, x='era', y='compound')
plt.title("Sentiment Spread by Era")
plt.show()
```

As indicated by the above boxplot and similarity in variance values between both eras, there likely is not a difference in homogeneity between the two eras; they seem to emulate the lack of significant difference in mean and aggregate counts of positive and negative sentiments showed in the previous section. We can continue to test the robustness of this conclusion via performing a test on the density of sentiment distributions by era.

```python
sns.kdeplot(data=genius_df, x='compound', hue='era', fill=True)
plt.title("Sentiment Distribution by Era")
plt.show()
```

As shown by the overlap between the two density maps, there does not appear to be much difference between era homogeneity, which supports the conclusion drawn by the previous figure. As a final, conclusive test of this hypothesis, we compare the standard deviations across both eras and perform a Levene test, which aims to test the null hypothesis that the variance between two examined groups is equal.

```python
pre = genius_df[genius_df['era'] == 'pre']['compound']
post = genius_df[genius_df['era'] == 'post']['compound']

# Standard deviations
pre_std = pre.std()
post_std = post.std()

print("Pre std:", pre_std)
print("Post std:", post_std)
```

There is an evidently marginal difference in standard deviations between eras, which further supports a lack in homogeneity differnces. 

```python
from scipy.stats import levene

pre = genius_df[genius_df['era'] == 'pre']['compound']
post = genius_df[genius_df['era'] == 'post']['compound']

levene(pre, post)
```

Supporting the minimal homogeneity differences shown in the figures, the variance values, and the standard deviations, the Levene test yielded a p-value of 0.62. As this value is well above conventional significance thresholds, we fail to reject the null hypothesis of equal variances. This provides strong evidence that the variability of lyrical sentiment—and thus the degree of homogeneity—has not meaningfully changed between the pre-streaming era of the early 2000s and the modern streaming era. 

At a broader level, this finding challenges our hypothesis that the rise of streaming platforms and algorithmic curation of music profiles has led to greater uniformity in popular music. Despite significant shifts in how music is distributed and consumed, the emotional diversity of top-charting songs appears largely unchanged. This stability may reflect enduring industry dynamics such as consistent listener preferences, the influence of major labels, or the nature of musical trends that continue to shape the emotional landscape of popular music regardless of technological change. However, it must be considered that the VADER test used to classify sentiment only groups lyrics into "positive" and "negative" categories, which does not represent the entire spectrum of emotions often expressed in music lyrics. The VADER tool was also initially designed to analyze the sentiment of social media posts, which makes it much worse at analyzing metaphorical or symbolic elements that are extremely prevalent in song lyrics. It also does not correctly compute sentiment for non-English lyrics. Thus, although this test concluded that, based on our data, there is not a statistically significant difference in emotional homogeneity across eras, it may be possible that a different or more specific sentiment analysis would reflect different results.

More generally, our analysis and results across musical and lyrical features both highlight that structural changes in the distribution of music do not necessarily translate to changes in creativity and diversity at the aggregate level. While streaming has transformed access and exposure, it may not have fundamentally altered the range of expression that resonates with mainstream audiences. Instead, the features that define mainstream appeal appear to persist over time, pointing to enduring patterns in listener preferences and industry production norms. The features that did prove statistically significant in difference across the eras, such as energy, loudness, and explicitness, had negligible correlation to the era variable. Even though such factors did change, likely due to evolving technology and societal norms, it was not to an impactful enough degree to claim that there is a measurable difference in musical diversity from before and after streaming became popularized. Thus, we conclude that our hypothesis about post-streaming songs being less diverse cannot be sufficiently supported, as a majority of our analysis and testing proved that the homogeneity between the two eras is negligibly different in terms of the variables we examined.
