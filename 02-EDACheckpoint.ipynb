# Research Question

## Background and Prior Work

As digitalization has rapidly advanced, no stone has been left unturned. The way we listen to music has been one of the biggest changes; progressing from physical cassettes, records, and CDs to singular devices that hold hundreds of millions of tracks. This unparalleled access to music has not come lightly. Beginning in the 2010s, music consumers are able to purchase or subscribe to a streaming platform to gain access to a dataset of songs, readily available to them. Since the release of streaming platforms like Spotify and Apple Music, the barrier of entry to consume and release music has lowered exponentially.

However, the way we consume music has also changed, with the algorithm becoming the ruling factor for many streaming platforms and social media sites. The more a song appeals to an algorithm, the more streams a song has, and all the more money it makes. This cause and effect has led to what we believe has been a homogenization of music since the streaming age began, where songs are designed for mass appeal. In this project, we’d like to explore whether or not music has become more homogenized when comparing post-streaming (2013-2026) and pre-streaming (2000-2013) eras. Our definition of homogeneity will include four features: energy, valence, acousticness, instrumentalness. Energy measures the "danceability" and animation of a given song, faster tempos and louder volumes connote higher scales of danceability and activity in a track, whereas softer, quieter volumes and slower tempos connote lower scales of energy. Valence describes the musical positivity, higher scales of valence include upbeat, happier, and euphoric music while lower values will include more moody, sadder, and tense musical qualities. High acousticness will include songs that sound organically produced physically by humans with instruments and lower values of acousticness will include songs that sound electronically produced, including techno-music. For example, more electronic music highlights faster tempos, which will be measured as in beats per minute (BPM). Instrumentalness highlights how many words are in a given track. A song with many lyrics will be graded lower on the instrumentalness scale, whereas tracks with mostly instrumental portions as very few lyrics and vocals will be graded higher on the scale. These four features: energy, valence, acousticness, and instrumentalness, are indicators of relative homogeneity. In our project, we would like to see if these features converge or diverge among the monthly top three Billboard Hot 100 songs during the streaming era (2013-2026) relative to the pre-streaming era (200-2012).

Perhaps one of the clearest signs of how music has changed is the change in the average length of music over time, which has decreased by more than 30 seconds over time. Many of the top songs of the past few years have clocked in at under three minutes, a clear departure from longer music. This is most likely tied to the difference streaming makes when considering song structure. Researchers have found a correlation between skipping music and song structure where users are likely to skip music before the song has even made it to 30 seconds. Since streaming services only pay artists for “plays” and “plays” only count if a song has been played for longer than 30 seconds, it stands to reason that many artists and songwriters have focused their energy towards creating catchy music that’s interesting enough to capture attention, but short enough that listeners won’t get bored and skipping, leaving artists without their hard-earned revenue.

In a previous study from Tilburg University, the impact of streaming services on homogenized mainstream music is mixed. In their research, Knox and Datta look at the supply and demand side of the industry, focusing on the “acoustic diversity” of a given song: the pitch, loudness, valence, tempo, and duration. In the early beginnings of platforms like Napster, an illegal music sharing platform, YouTube, and Spotify, artists were able to easily create and distribute their music to a large audience in a way that had not been seen before. Musicians tended to mimic each other in the early stages of the new widespread distribution of music, leading to an initial decrease in diversity in music quality. However, when streaming services like Spotify became more popularized, algorithms in place were able to provide new recommendation systems and search tools that enabled listeners to discover a vast amount of music readily available to them. From a demand standpoint, consumers have more choices of music selection available to them, thus allowing for more diversity within their consumption.

Despite the initial dip in acoustic diversity, there has been a weak, steady increase in acoustic diversity since the release of streaming platforms due to the large catalog of recorded music available. However, our project is looking to see if special media platforms like TikTok or Instagram influence the homogeneity of music selections within streaming platforms. We aim to explore this topic by focusing on the categorization of songs through factors such as tempo, lyrics, etc, that were on the Billboard Top 100 over the last 60-70 years. Through our exploration and analysis of this data, we hope to discover whether streaming has truly homogenized popular music or if streaming has had little to no effect on the variability of popular music.

# Hypothesis


We hypothesize that songs from the years 2013-2026, which would be considered the streaming era, will exhibit less variability in measurable factors such as danceability, energy valence, acousticness, instrumentalness, tempo, duration, loudness, key/mode, explicitness, and speechness when comparing them to songs from 2000-2012, which would be the pre-streaming era. This would indicate that popular songs over the years have become more homogenized. This could be for a wide variety of reasons, but we theorize that it is largely because of two reasons. The first is because of corporate involvement with music creation. Music corporations want to maximize profits and popularity of an artist's music, which makes them push for their artists to create music that is broader, more generic, and more formulaic, which has led popular songs to sound more similar in recent years. This is compounded by the second cause, the rise of streaming services. Streaming services are operated as a business that pits artists against each other in a popularity contest. More streams mean an artist's music is pushed out more, which means more money for the artist and exponentially more for the business. This also encourages streaming services to promote music that appeals to the broadest audience, as it generates the most revenue for them.

### Dataset #1 :BillBoard Hot100: Top 50 Annually

Scraped from Wikipedia: https://en.wikipedia.org/wiki/Billboard_Year-End_Hot_100_singles_of_1958

This dataset comprises of Billboard's annual overall ranking of the top 50 songs since 1958, including song title, artist, and rank (from 1-50). This dataset is the most up-to-date one of BillBoard chart data, but instead of going up to January/February 2026, its last recorded data ends at December of 2025. For the purposes of this data analysis, however, we will generalize the conclusions from song post 2014 to October 2025 to be part of our "post-streaming (2013-2026)" dataset, due to the large size of the dataset compared to the relatively small time period of missing values. 

The dataset has been filtered below for the appropriate years (2000-2025), and divided into separate dataframes sorted based on year and chart rank, taking the top 10 out the top 50 as our sample space. We chose to analyze starting from 2000 instead of from the inception of the Billboard Top 100 chart in 1958 because it would normalize the number of data points we have from our pre-streaming and post-streaming eras. 

```python
# run pip install lxml if not already
```

```python
import requests
import pandas as pd
import time
from io import StringIO

start_year = 2000
end_year = 2025

all_years = []

headers = {
    "User-Agent": "Mozilla/5.0"
}

for year in range(start_year, end_year + 1):
    try:
        url = f"https://en.wikipedia.org/wiki/Billboard_Year-End_Hot_100_singles_of_{year}"
        
        response = requests.get(url, headers=headers)
        response.raise_for_status()

        html = StringIO(response.text)
        tables = pd.read_html(html)

        for table in tables:
            if "Title" in table.columns or "Song" in table.columns:
                df = table
                break

        df = df.head(10).copy()

        df = df.iloc[:, -2:]
        
        df.columns = ["song", "artist"]
        
        df["year"] = year
        
        df = df[["year", "song", "artist"]]
        
        df.columns = [str(col).lower() for col in df.columns]

        if "title" in df.columns:
            df = df.rename(columns={"title": "song"})
        if "artist(s)" in df.columns:
            df = df.rename(columns={"artist(s)": "artist"})

        if "song" not in df.columns:
            df.rename(columns={df.columns[1]: "song"}, inplace=True)
        if "artist" not in df.columns:
            df.rename(columns={df.columns[2]: "artist"}, inplace=True)

        all_years.append(df)

        print(f"Finished {year}")

    except Exception as e:
        print(f"Skipped {year}: {e}")

final_df = pd.concat(all_years, ignore_index=True)

print(final_df.head())
```

```python
#artist cleanup (will be applied to Spotify 160k later as well)
import re

def normalize_artists(artist_string):
    if pd.isna(artist_string):
        return artist_string
    artist_string = artist_string.lower()
    artist_string = re.sub(r'featuring|feat\.?|with|&| x | and ', ',', artist_string) 
    artist_string = re.sub(r'\s*,\s*', ',', artist_string)
    artist_string = re.sub(r',+', ',', artist_string)
    artist_list = artist_string.split(',')
    artist_list = [a.strip() for a in artist_list if a.strip()]
        
    return ",".join(artist_list)
```

https://github.com/fcpercival/spotify-data-sorted

The 160K Spotify Songs (1921–2020) is a large music metadata collection derived from Spotify’s Web API. It contains roughly 160,000 tracks and includes artist information, genre labels, release year, and several audio features.

The most important metrics are Spotify’s audio features, which are standardized numeric descriptors. Danceability (unitless, 0–1) estimates how suitable a track is for dancing. Values closer to 1 indicate more danceable songs. Energy (0–1) measures intensity and activity; higher values correspond to louder and faster tracks. Valence (0–1) describes musical positivity. Higher values sound happier or more euphoric, while lower values tend to sound sadder or more tense. Acousticness (0–1) estimates the likelihood a track is acoustic rather than electronically produced. Instrumentalness (0–1) predicts whether a track contains vocals; values above ~0.5 suggest mostly instrumental music. These features are probabilities or model outputs rather than physical measurements, so they are relative indicators rather than exact physical quantities.

Several features are expressed in real physical units. Tempo is measured in beats per minute (BPM); most contemporary pop songs fall roughly between 90–130 BPM, while extreme values (e.g., >200 BPM) indicate very fast genres like speed metal or certain electronic styles. Duration is given in milliseconds; dividing by 1000 converts to seconds. Loudness is measured in decibels (dB), typically ranging from about −60 dB (very quiet) to 0 dB (maximum digital level). Key is encoded as an integer 0–11 representing the 12 pitch classes (C through B), and mode indicates major (1) or minor (0). Speechiness (0–1) measures the presence of spoken words; values above ~0.66 are likely spoken word (e.g., podcasts, rap verses heavy in speech), while values below ~0.33 are more purely musical. These metrics allow analysis of musical style, mood, production trends, and historical changes over time.

The primary concern with bias in this dataset is derived from the fact that the data is taken from Spotify’s catalog and recommendation algorithms, meaning it reflects Spotify’s licensing agreements and popularity trends, not a perfectly representative sample of all music ever recorded. Earlier decades (e.g., 1920s–1950s) are likely underrepresented and skewed toward recordings that have been digitized and re-released, which may also show biases.
```python
import pandas as pd
import re
import ast

def clean_title(title):
    if pd.isna(title):
        return title
    
    title = title.lower()
    
    # remove quotes
    title = title.replace('"', '')
    title = title.replace("'", "")
    
    # remove parentheses content
    title = re.sub(r"\(.*?\)", "", title)
    
    # remove dash versions
    title = re.sub(r"-.*", "", title)
    
    # remove punctuation
    title = re.sub(r"[^\w\s]", "", title)
    
    return title.strip()

def clean_spotify_artists(artist_string):
    if pd.isna(artist_string):
        return None
    
    # convert string list → python list
    artist_list = ast.literal_eval(artist_string)
    
    # lowercase and strip
    artist_list = [a.lower().strip() for a in artist_list]
    
    return ",".join(artist_list)
```

### Dataset #3: Genius Lyrics Dataset

For this project, we sourced lyrics from the Genius Lyrics Database, filtering entries to match our selected sample of Billboard songs. The Genius database is a crowdsourced repository containing over 8 million lyrics, making it the largest publicly available resource for lyrical and musical knowledge. Contributions are often verified by the artists themselves, ensuring a high degree of accuracy not only for the lyrics but also for the annotations and analyses associated with each song. Using the LyricsGenius Python client, we retrieved lyrics corresponding to our sample of songs from both the pre- and post-streaming eras, enabling a consistent and structured dataset for analysis. These lyrics form the basis for our exploration of lyrical features, including sentiment, mood, and thematic content. In particular, we utilize a pretrained machine learning model provided by Nyckel, which classifies lyrics according to emotional tone and positivity/negativity, allowing us to quantitatively compare lyrical trends across different time periods.

```python
#run pip install lyricsgenius first if you do not have it
import lyricsgenius

GENIUS_TOKEN = os.getenv("GENIUS_API_TOKEN")

if GENIUS_TOKEN is None:
    raise ValueError(
        "Genius API token not found. Please set GENIUS_API_TOKEN as an environment variable."
    )

genius = lyricsgenius.Genius(
    GENIUS_TOKEN,
    skip_non_songs=True,
    remove_section_headers=True,
    excluded_terms=["(Remix)", "(Live)"]
)

print("Genius client initialized successfully.")
```

```python
def clean_artist(artist):
    return artist.split(",")[0].strip()
def clean_title(title):
    title = re.sub(r"\(.*?\)", "", title)  
    title = re.sub(r"- .*", "", title)     
    return title.strip()
    
pre_streaming_data = pd.read_csv("data/02-processed/pre_streaming_top10_annual")
post_streaming_data = pd.read_csv("data/02-processed/post_streaming_top10_annual")

pre_streaming_data["genius_artist"] = pre_streaming_data["artists"].apply(clean_artist)
pre_streaming_data["genius_title"] = pre_streaming_data["song"].apply(clean_title)

post_streaming_data["genius_artist"] = post_streaming_data["artists"].apply(clean_artist)
post_streaming_data["genius_title"] = post_streaming_data["song"].apply(clean_title)
```

```python
pre_streaming_data
```

```python
from concurrent.futures import ThreadPoolExecutor, as_completed
import pandas as pd

def fetch_lyrics(row, era):
    title = row["genius_title"]
    artist = row["genius_artist"]

    try:
        song = genius.search_song(title, artist, get_full_info=False)

        return {
            "title": row["genius_title"],
            "artist": row["artists"],
            "lyrics": song.lyrics if song else None,
            "era": era
        }

    except Exception as e:
        print(f"Error with {title}: {e}")
        return {
            "title": row["genius_title"],
            "artist": row["artists"],
            "lyrics": None,
            "era": era
        }


lyrics_rows = []

# combine datasets so we only parallelize once
tasks = []

for _, row in pre_streaming_data.iterrows():
    tasks.append((row, "pre"))

for _, row in post_streaming_data.iterrows():
    tasks.append((row, "post"))

# run parallel requests
with ThreadPoolExecutor(max_workers=10) as executor:
    futures = [executor.submit(fetch_lyrics, row, era) for row, era in tasks]

    for future in as_completed(futures):
        lyrics_rows.append(future.result())

# create dataframe
final_df = pd.DataFrame(lyrics_rows)
```

### Musical Feature Variability Across Pre-streaming and Post-streaming Eras

To evaluate whether the variability of mainstream music has changed between the pre-streaming era (2000–2012) and the streaming era (2013–2026), we compared the distributions of several features, including acousticness, danceability, energy, loudness, speechiness, tempo, valence, and duration. For categorical variables, we analyzed key, explicit content, and musical mode.

Energy increased significantly in the streaming era (t = 2.85, p = 0.0049, Cohen’s d = 0.387), indicating that songs in the streaming era tend towards having higher  intensity. Similarly, loudness increased significantly (t = 2.35, p = 0.0199, Cohen’s d = 0.323), indicating that modern popular songs are produced with louder levels. These results are consistent with the boxplots, which show an upward shift in the distributions of both energy and loudness after 2013, or rather, after the pre-streaming era.

Song duration also changed substantially between eras. Duration decreased significantly in the streaming era with t = 5.06, p < 0.000001, and Cohen’s d = 0.696, representing the largest observed effect size among the tested variables. This large effect suggests a meaningful structural change in song length, likely reflecting the incentive in streaming platforms like Spotify, where shorter songs can lead to more plays and therefore, more revenue for the artist.

In contrast, several other features don't show much evidence of change. Danceability (t = 0.196, p = 0.845, d = 0.026), tempo (t = −0.400, p = 0.689, d = −0.054), valence (t = 0.456, p = 0.649, d = 0.062), and speechiness (t = −1.583, p = 0.115, d = −0.215) exhibit highly overlapping distributions between eras. Acousticness shows a minor difference (t = −1.90, p = 0.059), but the effect size remains relatively small (d = −0.264).

For categorical features, chi-square tests were used to evaluate whether distributions changed between eras. The proportion of songs marked as explicit increased significantly in the streaming era (χ² = 9.00, p = 0.0027), suggesting a greater prevalence of explicit lyrical content in modern popular music. However, the distributions of musical key (χ² = 13.03, p = 0.291) and mode (χ² = 0.013, p = 0.910) show no statistically significant differences across eras, indicating that tonal structure has remained stable.

### Musical Diversity Score

In order to measure the quantity of musical characteristics, we used the standardized Spotify audio features for each song in order to compute a musical diversity score between each song. Using the song traits: acousticness, danceability, energy, instrumentalness, speechiness, tempo, valence, loudness, and duration, we measure each song from the top 5 charts from before and after streaming era, and were able to calculate how diverse, if at all, songs were from each other. We then compared the diversity score between songs from the pre-streaming era and from the streaming era. Comparing both periods, we see that the mean diversity score for the pre-streaming era is (0.7098) and the streaming era is (0.7085), which shows that the mean score for the two eras are almost the same.

 To confirm the results from the model, we used a Welch two-sample t-test and since p is greater than 0.5,(p = 0.946), it tells us that the results are not statistically significant. When we compare the music diversity score from the top Billboard songs from both eras, the songs have remained similar in traits and show a lack of diversity. 
 
Finalizing our analysis from the music diversity scores, we are able to determine that some music traits like loudness, energy, and song duration have evolved differently between pre-streaming era and current era. However, the evolution of the songs from both eras is minor, and these mainstream songs from the top Billboard charts lack diversity and have remained constant.

### Summary of Findings

Overall, the exploratory data analysis suggests that while certain individual musical attributes have shifted in the streaming era, such as increased energy, louder production, shorter song durations, and a higher amount of explicit lyrics, these changes do not lead into a measurable difference in the overall variability of musical features among top Billboard songs.

The diversity score analysis provides strong statistical evidence that the overall level of musical variability has remained stable between the pre-streaming and streaming eras. With nearly identical means (0.7098 vs. 0.7085) and a non-significant Welch t-test (p = 0.946), the results suggest that although the stylistic characteristics of popular music have evolved, the overall diversity of musical features within mainstream hits has not significantly increased or decreased since the rise of streaming platforms.

```python
#Cohen's d to measure effect size
def cohens_d(x, y):
    
    nx = len(x)
    ny = len(y)
    
    pooled_std = np.sqrt(((nx-1)*np.var(x) + (ny-1)*np.var(y)) / (nx+ny-2))
    
    return (np.mean(x) - np.mean(y)) / pooled_std
```

```python
def cramers_v(table):
    
    chi2 = chi2_contingency(table)[0]
    n = table.sum().sum()
    
    r, k = table.shape
    
    return np.sqrt(chi2 / (n * (min(r-1, k-1))))
```

### Computing Overall Feature Diversity for Pre-streaming VS. Post-streaming Era (Continuous Features Only)

```python
diversity_features = [
    "acousticness",
    "danceability",
    "energy",
    "loudness",
    "speechiness",
    "tempo",
    "valence",
    "duration_min"
]
```

```python
merged_df["diversity_score"] = merged_df[diversity_features].abs().mean(axis=1)
pre_div = merged_df[merged_df["era"] == 'pre']
post_div = merged_df[merged_df["era"] == 'post']
```

```python
#Boxplot of Musical Diversity by Era
plt.figure(figsize=(6,5))

sns.boxplot(
    data=merged_df,
    x="era",
    y="diversity_score"
)

plt.title("Musical Diversity Score by Era")
plt.show()
```

```python
#Density Plot of Distribution of Musical Diversity Across Eras
plt.figure(figsize=(7,5))

sns.kdeplot(pre_div["diversity_score"], label="Pre-streaming", fill=True)
sns.kdeplot(post_div["diversity_score"], label="Post-streaming", fill=True)

plt.title("Distribution of Musical Diversity Scores")
plt.legend()
plt.show()
```

```python
#Testing normality of results
stats.shapiro(pre_div["diversity_score"])
stats.shapiro(post_div["diversity_score"])
```

```python
#p > 0.05 - results are normal. Using the Welch t-test for normal data:
stats.ttest_ind(
    pre_div["diversity_score"],
    post_div["diversity_score"],
    equal_var=False
)
```

```python
#Effect size of diversity scores
def cohens_d(x,y):
    pooled_sd = np.sqrt((x.var() + y.var()) / 2)
    return (x.mean() - y.mean()) / pooled_sd

cohens_d(pre_div["diversity_score"], post_div["diversity_score"])
```
