# Dataset #1: 160K Spotify Songs (1921-2020)

The 160K Spotify Songs (1921–2020) is a large music metadata collection derived from Spotify’s Web API. It contains roughly 160,000 tracks spanning nearly a century of recorded music and includes artist information, genre labels, release year, and a wide range of audio features. The dataset is structured as a tabular CSV file, where each row represents a song and each column represents either metadata (e.g., artist, track name, year) or a quantitative audio feature.

The most important metrics are Spotify’s audio features, which are standardized numeric descriptors. Danceability (unitless, 0–1) estimates how suitable a track is for dancing based on tempo, rhythm stability, beat strength, and regularity; values closer to 1 indicate more danceable songs. Energy (0–1) measures intensity and activity; higher values correspond to louder, faster, and more “energetic” tracks. Valence (0–1) describes musical positivity; higher values sound happier or more euphoric, while lower values tend to sound sadder or more tense. Acousticness (0–1) estimates the likelihood a track is acoustic rather than electronically produced. Instrumentalness (0–1) predicts whether a track contains vocals; values above ~0.5 suggest mostly instrumental music. These features are probabilities or model outputs rather than physical measurements, so they are relative indicators rather than exact physical quantities.

Several features are expressed in real physical units. Tempo is measured in beats per minute (BPM); most contemporary pop songs fall roughly between 90–130 BPM, while extreme values (e.g., >200 BPM) indicate very fast genres like speed metal or certain electronic styles. Duration is given in milliseconds; dividing by 1000 converts to seconds. Loudness is measured in decibels (dB), typically ranging from about −60 dB (very quiet) to 0 dB (maximum digital level). Key is encoded as an integer 0–11 representing the 12 pitch classes (C through B), and mode indicates major (1) or minor (0). Speechiness (0–1) measures the presence of spoken words; values above ~0.66 are likely spoken word (e.g., podcasts, rap verses heavy in speech), while values below ~0.33 are more purely musical. These metrics allow analysis of musical style, mood, production trends, and historical changes over time.

The primary concern with bias in this dataset is derived from the fact that the data is taken from Spotify’s catalog and recommendation algorithms, meaning it reflects Spotify’s licensing agreements and popularity trends, not a perfectly representative sample of all music ever recorded. Earlier decades (e.g., 1920s–1950s) are likely underrepresented and skewed toward recordings that have been digitized and re-released, which may also show biases . Overall, the dataset is extremely useful for large-scale music analytics, but interpretations should consider algorithmic bias, catalog selection bias, and temporal imbalance.
