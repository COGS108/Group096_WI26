## Research Question

To what extent has the variability in song duration and genre diversity among the monthly top three Billboard Hot 100 songs decreased in the streaming era (2013-2026) relative to the pre-streaming era (2000-2012)?
 
(We chose this time epoch because 2013 was the beginning of the streaming-dominant era, as it generally marks the first year in which streaming became the primary mode of music consumption in the United States. To allow for a balanced comparison, we chose to analyze an equal span of time before (2000-2012) and after (2013-2026) this shift.)

## Hypothesis


We hypothesize that songs from the years 2013-2026, which would be considered the streaming era, will exhibit less variability in measurable factors such as danceability, energy valence, acousticness, instrumentalness, tempo, duration, loudness, key/mode, explicitness, and speechness when comparing them to songs from 2000-2012, which would be the pre-streaming era. This would indicate that popular songs over the years have become more homogenized. This could be for a wide variety of reasons, but we theorize that it is largely because of two reasons. The first is because of corporate involvement with music creation. Music corporations want to maximize profits and popularity of an artist's music, which makes them push for their artists to create music that is broader, more generic, and more formulaic, which has led popular songs to sound more similar in recent years. This is compounded by the second cause, the rise of streaming services. Streaming services are operated as a business that pits artists against each other in a popularity contest. More streams mean an artist's music is pushed out more, which means more money for the artist and exponentially more for the business. This also encourages streaming services to promote music that appeals to the broadest audience, as it generates the most revenue for them.

# Dataset #1: 160K Spotify Songs (1921-2020)

https://github.com/fcpercival/spotify-data-sorted

The 160K Spotify Songs (1921–2020) is a large music metadata collection derived from Spotify’s Web API. It contains roughly 160,000 tracks and includes artist information, genre labels, release year, and several audio features.

The most important metrics are Spotify’s audio features, which are standardized numeric descriptors. Danceability (unitless, 0–1) estimates how suitable a track is for dancing. Values closer to 1 indicate more danceable songs. Energy (0–1) measures intensity and activity; higher values correspond to louder and faster tracks. Valence (0–1) describes musical positivity. Higher values sound happier or more euphoric, while lower values tend to sound sadder or more tense. Acousticness (0–1) estimates the likelihood a track is acoustic rather than electronically produced. Instrumentalness (0–1) predicts whether a track contains vocals; values above ~0.5 suggest mostly instrumental music. These features are probabilities or model outputs rather than physical measurements, so they are relative indicators rather than exact physical quantities.

Several features are expressed in real physical units. Tempo is measured in beats per minute (BPM); most contemporary pop songs fall roughly between 90–130 BPM, while extreme values (e.g., >200 BPM) indicate very fast genres like speed metal or certain electronic styles. Duration is given in milliseconds; dividing by 1000 converts to seconds. Loudness is measured in decibels (dB), typically ranging from about −60 dB (very quiet) to 0 dB (maximum digital level). Key is encoded as an integer 0–11 representing the 12 pitch classes (C through B), and mode indicates major (1) or minor (0). Speechiness (0–1) measures the presence of spoken words; values above ~0.66 are likely spoken word (e.g., podcasts, rap verses heavy in speech), while values below ~0.33 are more purely musical. These metrics allow analysis of musical style, mood, production trends, and historical changes over time.

The primary concern with bias in this dataset is derived from the fact that the data is taken from Spotify’s catalog and recommendation algorithms, meaning it reflects Spotify’s licensing agreements and popularity trends, not a perfectly representative sample of all music ever recorded. Earlier decades (e.g., 1920s–1950s) are likely underrepresented and skewed toward recordings that have been digitized and re-released, which may also show biases. Furthermore, the data does not consist of music from 2020-2026, which is a large segment of the time period we are analyzing for post-streaming.

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("data/00-raw/music-data.csv")
drops = ['liveness', 'popularity']
df = df.drop(columns=drops)
print(df.columns)
print(df.isnull().sum())

df["duration_min"] = df["duration_ms"] / 60000
df = df.drop(columns=["duration_ms"])

features = [
    "acousticness",
    "danceability",
    "energy",
    "instrumentalness",
    "loudness",
    "speechiness",
    "tempo",
    "valence",
    "duration_min"
]

#normalize data based on difference from mean
#eg: 0 = average tempo, +1 = 1 std above mean, −1 = 1 std below mean
X =  df[features].copy()
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_scaled_df = pd.DataFrame(X_scaled, columns=features)
df[features] = X_scaled_df

df.head()
```

### Dataset #3: BillBoard Hot100 Dataset

https://www.kaggle.com/datasets/ludmin/billboard?select=hot100.csv

This dataset comprises comprehensive data about every song that has ever charted on the BillBoard Hot100 (since 1958), including song title, artist, rank (from 1-100), position peaked on the Hot100 (from 1-100), number of weeks on the Hot 100, date the song got ranked on the chart, and the song's release date. This dataset is the most up-to-date one of BillBoard chart data, but instead of going up to January/February 2026, its last recorded data ends at October 2025. For the purposes of this data analysis, however, we will generalize the conclusions from song post 2014 to October 2025 to be part of our "post-streaming (2013-2026)" dataset, due to the large size of the dataset compared to the relatively small time period of missing values. 

The dataset has been filtered below for the appropriate years (2000-2026), and divided into separate dataframes (hot100_2000_2013_sorted, and hot100_2014_2026_sorted) sorted based on year, month, and chart rank. It contains 2190 data points for songs charted between 2000-2013, and 1851 data points for songs charted between 2014-2026. This accounts for the fact that more than three songs chart in the top three ranks in any given month. There are no missing or null values recorded for any songs.

```python
!pip install kagglehub[pandas-datasets]
import kagglehub

kagglehub.dataset_download(
    "ludmin/billboard",
    path="hot100.csv",
    output_dir="./data/00-raw"
)
```

```python
import pandas as pd
hot100 = pd.read_csv("data/00-raw/hot100.csv")
hot100["release_date"] = pd.to_datetime(hot100["Date"])

hot100["year"] = hot100["release_date"].dt.year
hot100["month"] = hot100["release_date"].dt.month

hot100_filtered = hot100[
    (hot100["year"].between(2000, 2026)) &
    (hot100["Rank"].between(1, 3))
]
display(hot100_filtered.shape)
display(hot100.isna().sum())
```

```python
hot100_2000_2013 = hot100_filtered[hot100_filtered["year"] <= 2013]
hot100_2000_2013.shape
```

```python
hot100_2014_2026 = hot100_filtered[hot100_filtered["year"] >= 2014]
hot100_2014_2026.shape
```

```python
hot100_2000_2013_sorted = hot100_2000_2013.sort_values(
    by=["year", "month", "Rank"]
)
hot100_2000_2013_sorted
```

```python
hot100_2014_2026_sorted = hot100_2014_2026.sort_values(
    by=["year", "month", "Rank"]
)
hot100_2014_2026_sorted
```

```python
hot100_2000_2013_sorted.to_csv("data/02-processed/bb_hot100_2000_2013.csv", index=False)
hot100_2014_2026_sorted.to_csv("data/02-processed/bb_hot100_2014_2026.csv", index=False)
```
